```python
# Here we smooth the numerical results for folded configuration
# 3-point smoothing with numpy
from scipy import ndimage

# Smooth with a 3-point moving average
#N0_numerical_fold = ndimage.uniform_filter1d(N0_numerical_fold, size=20)

from scipy import signal

# The elegant solution that handles edges
def smooth_with_edges(data, window_size):
    kernel = np.ones(window_size, dtype=np.float128)
    numerator = signal.convolve(data, kernel, mode='same')
    denominator = signal.convolve(np.ones(len(data)), kernel, mode='same')
    return numerator / denominator

# Don't smooth the first and last 2 points
N0_smooth_interior = smooth_with_edges(N0_numerical_fold[60:], 10)
N0_numerical_fold_smooth = np.concatenate([N0_numerical_fold[:60], N0_smooth_interior])
```

```python
from scipy.ndimage import gaussian_filter1d

# Smooth everything first
smoothed_all = gaussian_filter1d(N0_numerical_fold, sigma=1.5)

# Start with original data
N0_numerical_fold_smooth = N0_numerical_fold.copy()

# Gradually blend from original to smoothed over ~10 points
transition_start = 18
transition_length = 10
for i in range(transition_length):
    weight = i / transition_length  # 0 to 1
    idx = transition_start + i
    if idx < len(N0_numerical_fold):
        N0_numerical_fold_smooth[idx] = (1-weight) * N0_numerical_fold[idx] + weight * smoothed_all[idx]

# Fully smoothed after transition
N0_numerical_fold_smooth[transition_start + transition_length:] = smoothed_all[transition_start + transition_length:]
```
